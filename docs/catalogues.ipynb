# Catalogues and OSM extracts

`fetch` runs the whole pipeline, but every stage is also available on its
own. This page searches the Mobility Database catalogue directly and
downloads OpenStreetMap data for an area.

## Search the Mobility Database

The [Mobility Database](https://mobilitydatabase.org/) catalogues GTFS feeds
worldwide. Let's search it for the feeds whose bounding box meets central
Turku:

````python
import transitio

db = transitio.MobilityDatabase()                 # catalog client
feeds = db.search_feeds(aoi=(22.20, 60.42, 22.32, 60.48))
for feed in feeds:
    print(feed.id, feed.provider, feed.official)
````

Without an API token, the client reads the catalogue's public CSV export and
warns about it. Notice that a search by bounding box also finds national and
international feeds whose extent reaches the area; the feed index, which
[Choosing feeds](choosing_feeds.ipynb) uses, knows which feeds serve each
place.

`download_latest` downloads a feed's newest dataset:

````python
path = db.download_latest(feeds[0])
path.name
````

With an API token (the `MOBILITY_API_REFRESH_TOKEN` environment variable),
the client can also pick the dataset version that serves a day and read the
hosted validation report:

```python
dataset = db.dataset_for(feeds[0], when="2026-09-01")
path = db.download(dataset)                        # cached, checksum-verified
hosted = db.validation_report(dataset)             # canonical-validator report
```

## Download OpenStreetMap data

`fetch_pbf` downloads the smallest OpenStreetMap extract that contains an
area, from Geofabrik, BBBike or Movisda, and crops it to the area's bounding
box:

````python
pbf = transitio.fetch_pbf((22.20, 60.42, 22.32, 60.48))
pbf.name
````
